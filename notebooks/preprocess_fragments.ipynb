```python
HERE = Path(_dh[-1])
binders_file = HERE / '..' / 'data' / 'temp' / 'docking_2' / 'results_filtered_chembl_kinases.sdf'
mols_binding = utils.read_sdf(binders_file)
```

```python
PATH_TO_LIB = HERE / '..' / 'data' / 'fragment_library_custom_filtered'
fragment_library = utils.read_fragment_library(PATH_TO_LIB)
fragment_library.keys()
```

```python
filtered_mols = []
for mol in mols_binding: 
    fragments = ast.literal_eval(mol.GetProp("smiles_fragments"))
    fragments_dummy = ast.literal_eval(mol.GetProp("smiles_fragments_dummy"))
    fragments["FP"] = fragments_dummy["FP"]
    for subpocket in fragments.keys():
        ind = fragment_library[subpocket].index[fragment_library[subpocket]['smiles'] == fragments[subpocket]]
        if len(ind) > 0: 
            print("here")
            frag_ids = subpocket + '_' + str(ind[0]) + ','
        else: 
            frag_ids = None
            break 
    if frag_ids is not None: 
        mol.SetProp("frag_ids", frag_ids[:-1])
        filtered_mols.append(mol)
```
